```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from statistics_backend import (
    clean_data, descriptive_statistics, bootstrap_confidence_interval,
    clt_simulation, one_sample_t_test, pearson_correlation, group_comparison
)

data = clean_data(pd.read_csv("coffee_data.csv"))
data.head()
```

```python
print("Descriptive statistics")
pd.DataFrame(descriptive_statistics(data), index=["Value"]).T
```

```python
confidence = 0.95
lower, upper = bootstrap_confidence_interval(data, confidence=confidence)
print(f"{confidence:.0%} bootstrap confidence interval: {lower:.3f} to {upper:.3f} cups/day")
```

```python
hypothesized_mean = 2.0
test = one_sample_t_test(data, hypothesized_mean=hypothesized_mean)
print(test)
if test["p_value"] < 0.05:
    print("Reject H0 at alpha = 0.05.")
else:
    print("Fail to reject H0 at alpha = 0.05.")
```

```python
sample_size = min(30, len(data))
sample_means = clt_simulation(data, sample_size=sample_size, repetitions=5000)

plt.figure()
plt.hist(sample_means, bins=30, edgecolor="black")
plt.axvline(data["Cups_Per_Day"].mean(), linestyle="--", label="Dataset mean")
plt.xlabel("Mean cups/day in repeated samples")
plt.ylabel("Frequency")
plt.title(f"Sampling distribution of means (n={sample_size})")
plt.legend()
plt.show()
```

```python
if {"Cups_Per_Day", "Money_Per_Week"}.issubset(data.columns):
    print("Pearson correlation")
    print(pearson_correlation(data))

if "Coffee_Type" in data.columns:
    print("One-way ANOVA")
    print(group_comparison(data))
```

## Interpretation reminders
- The null hypothesis for the t-test is that the population mean equals the chosen hypothesized mean.
- A p-value below 0.05 is evidence against the null under the test assumptions; it is not the probability that the null is true.
- A confidence interval describes uncertainty in an estimated population mean, assuming the sample design is appropriate.
- The dataset is synthetic. Replace it with collected survey data for conclusions about your target population.
